```python
import pygame
import random

class FlappyBirdEnv:
    def __init__(self):
        self.screen = pygame.display.set_mode((288, 512))
        self.clock = pygame.time.Clock()
        self.bird_y = 256
        self.bird_speed = 0
        self.pipe_x = 288
        self.pipe_gap = 100
        self.pipe_height = random.randint(50, 300)
        self.score = 0

    def reset(self):
        self.bird_y = 256
        self.bird_speed = 0
        self.pipe_x = 288
        self.pipe_height = random.randint(50, 300)
        self.score = 0
        return self.get_state()

    def get_state(self):
        return [self.bird_y, self.pipe_x, self.pipe_height]

    def step(self, action):
        if action == 1:
            self.bird_speed = -9
        self.bird_speed += 1
        self.bird_y += self.bird_speed
        self.pipe_x -= 3

        if self.pipe_x < -52:
            self.pipe_x = 288
            self.pipe_height = random.randint(50, 300)
            self.score += 1

        done = self.bird_y > 512 or self.bird_y < 0
        reward = 1 if not done else -100

        return self.get_state(), reward, done, {}

    def render(self):
        self.screen.fill((255, 255, 255))
        pygame.draw.rect(self.screen, (0, 255, 0), (self.pipe_x, 0, 52, self.pipe_height))
        pygame.draw.rect(self.screen, (0, 255, 0), (self.pipe_x, self.pipe_height + self.pipe_gap, 52, 512))
        pygame.draw.rect(self.screen, (255, 0, 0), (50, self.bird_y, 34, 24))
        pygame.display.update()
        self.clock.tick(30)

```

```python
from stable_baselines3 import PPO
from stable_baselines3.common.envs import DummyVecEnv
from stable_baselines3.common.vec_env import VecFrameStack

# Define environment
env = DummyVecEnv([lambda: FlappyBirdEnv()])

# PPO model
model = PPO("MlpPolicy", env, verbose=1)

# Train the model
model.learn(total_timesteps=100000)

# Save the model
model.save("flappy_bird_ai")
```
